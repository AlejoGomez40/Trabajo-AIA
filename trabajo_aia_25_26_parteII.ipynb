Creamos una lista con el nombre de las 5 clases. En castellano sería: margaritas, diente de león, rosas, girasoles y tulipanes.

Preparamos variables con las rutas de los diferentes directorios a crear:

```python
train_dir = os.path.join(base_dir, 'train')
val_dir = os.path.join(base_dir, 'val')

daisy_dir = os.path.join(train_dir, 'daisy')
dandelion_dir = os.path.join(train_dir, 'dandelion')
roses_dir = os.path.join(train_dir, 'roses')
sunflowers_dir = os.path.join(train_dir, 'sunflowers')
tulips_dir = os.path.join(train_dir, 'tulips')
```

```python
SPLIT_RATIO=0.8

for cl in classes:
    # path de las imagenes de la clase cl
    img_path = os.path.join(base_dir, cl)

    # obtenemos la lista de todas las imagenes
    images = glob.glob(img_path + '/*.jpg')
    print("{}: {} Imagenes".format(cl, len(images)))

    # determinamos cuantas imagenes son el 80%
    num_train = int(round(len(images)*SPLIT_RATIO))

    # separamos las imagenes en dos listas
    train, val = images[:num_train], images[num_train:]

    # creamos la carpeta de train/clase y val/clase
    if not os.path.exists(os.path.join(base_dir, 'train', cl)):
        os.makedirs(os.path.join(base_dir, 'train', cl))
    else:
        shutil.rmtree(os.path.join(base_dir, 'train', cl))

    if not os.path.exists(os.path.join(base_dir, 'val', cl)):
        os.makedirs(os.path.join(base_dir, 'val', cl))
    else:
        shutil.rmtree(os.path.join(base_dir, 'val', cl))

    for t in train:
        shutil.move(t, os.path.join(base_dir, 'train', cl))

    for v in val:
        shutil.move(v, os.path.join(base_dir, 'val', cl))
```

# c) Modelo 1: CNN básica (objetivo: accuracy_valid > 60%)


Implemente una red convolucional para resolver el problema de clasificación. Para ello se sugiere una CNN con 3 capas convolucionales + pooling con la siguiente estructura:

Bloque de procesamiento de imagen:
1.   32 kernels -> 64 kernels -> 96 kernels
2.   kernels de 3x3.
3.   Stride = 1 y padding = SI.
4.   Función de activación ReLU.
5.   Maxpooling de 2x2 con stride clásico de 2 pixeles.
6.   Igualamos el tamaño de todas las imágenes a 150 x 150.

Bloque de decisión:
7.   Capa densa de 512 neuronas.
8.   Capa densa de salida.

**Tarea 4: Define un modelo con la estructura anterior**



```python
model = keras.Sequential()

model.add(keras.layers.Conv2D(
              filters=32,         # filtros
              kernel_size=(3, 3), # tamaño filtro
              strides=(1, 1),     # como viaja, es el valor por defecto
              padding='same',    # utiliza padding
              activation='relu',
              name='conv1',
              input_shape=(150, 150, 3))) #Definimos el tamaño de entrada

model.add(keras.layers.MaxPooling2D(
              pool_size=(2, 2),   # max de cada bloque 2x2
              strides=(2, 2)))    # no overlapping con el pool size

model.add(keras.layers.Conv2D(
              filters=64,         # filtros
              kernel_size=(3, 3), # tamaño filtro
              strides=(1, 1),     # como viaja, es el valor por defecto
              padding='same',    # utiliza padding
              activation='relu',
              name='conv2'))

model.add(keras.layers.MaxPooling2D(
              pool_size=(2, 2),   # max de cada bloque 2x2
              strides=(2, 2)))    # no overlapping con el pool size

model.add(keras.layers.Conv2D(
              filters=96,         # filtros
              kernel_size=(3, 3), # tamaño filtro
              strides=(1, 1),     # como viaja, es el valor por defecto
              padding='same',    # utiliza padding
              activation='relu',
              name='conv3'))


model.add(keras.layers.MaxPooling2D(
              pool_size=(2, 2),   # max de cada bloque 2x2
              strides=(2, 2)))    # no overlapping con el pool size


model.add(tf.keras.layers.Flatten()) #Añadimos la capa de aplanado

model.add(keras.layers.Dense(units=512, activation='relu'))
model.add(keras.layers.Dense(units=5, activation='softmax'))
```

**Tarea 5: Indica el shape de la imagen antes y después de cada capa de la red. Explica cómo has obtenido dichos valores**

| Capa | Shape a la salida | #Parámetros | Explicación |
| :--- | :--- | :--- | :--- |
| Conv_1 | (150, 150, 32) | 896 | Aplica 32 filtros de 3x3 a la imagen de 3 canales. El tamaño no cambia porque hay Padding y slide de 1. Cálculo: (3x3x3 pesos + 1 vias) * 32. |
| Pool_1 | (75, 75, 32) | 0 | Reduce el tamaño a la mitad usando una capa de 2x2 y slide de 2. No aprende pesos. |
| Conv_2 | (75, 75, 64) | 18,496 | Aplica 64 filtros de 3x3 sobre los 32 capas de características de la capa anterior. El tamaño no cambia porque hay Padding y slide de 1. |
| Pool_2 | (37, 37, 64) | 0 | Vuelve a reducir el tamaño a la mitad (75/2 se trunca a 37). |
| Conv_3 | (37, 37, 96) | 55,392 | Extrae características aplicando 96 filtros de 3x3 sobre los 64 canales previos. El tamaño no cambia porque hay Padding y slide de 1.|
| Pool_3 | (18, 18, 96) | 0 | Última reducción (37/2 se trunca a 18). |
| Flatten | 31104 | 0 | Aplana el volumen 3D (18 * 18 * 96) en un vector 1D para conectarlo a la red neuronal clásica. |

# d) Modelo 2: reducción del overfitting (objetivo: accuracy_valid > 70%)

Para mejorar el accuracy del modelo, vamos a incorporar las dos técnicas más habituales de reducción del overfitting:

    * drop-out
    * data augmentation
    

**Tarea 9: Explica en que consisten y qué utilidad tienen para nuestro problema**





1. Drop-out
¿En qué consiste? En "apagar" aleatoriamente un porcentaje de neuronas durante el entrenamiento.

Utilidad: Evita que la red simplemente memorice las imágenes. Al forzar al modelo a no depender siempre de las mismas neuronas, aprende patrones más generales y reduce el overfitting.

2. Data Augmentation
¿En qué consiste? En crear "nuevas" imágenes durante el entrenamiento aplicando pequeñas modificaciones a las originales (rotar, hacer zoom, recortar, voltear).

Utilidad: Multiplica la cantidad de datos disponibles y le enseña a la red a reconocer el objeto sin importar su tamaño, posición o ángulo de visión.

**Tarea 10: Construye un nuevo modelo 2 incorporando (en el modelo 1 anterior) el dropout adecuado**

```python
#Esta vez creamos el modelo utilizando la API funcional y aprovechando los valores por defecto
img_input = keras.layers.Input(shape=(150, 150, 3))

x = keras.layers.Conv2D(32, 3, padding='same', activation='relu')(img_input)
x = keras.layers.MaxPooling2D(2)(x)
x = keras.layers.Conv2D(64, 3, padding='same', activation='relu')(x)
x = keras.layers.MaxPooling2D(2)(x)
x = keras.layers.Convolution2D(96, 3, padding='same', activation='relu')(x)
x = keras.layers.MaxPooling2D(2)(x)
x = keras.layers.Flatten()(x)
x = keras.layers.Dense(512, activation='relu')(x)
x = keras.layers.Dropout(0.5)(x) #Añadimos la capa de dropout
output = keras.layers.Dense(5, activation='softmax')(x)

model_2 = keras.Model(img_input, output)
```

**Tarea 11: Explica que tipos de augmentation vas a considerar y que utilidad tienen en nuestro problema de clasificacion**


Tipos de data augmentation a considerar:
*   Rotación: Podemos considerar una leve rotación en las imágenes de las flores, ya que seguirán siendo imágenes que tienen sentido. Pero tampoco podremos rotar en exceso. Ayudando al modelo así a reconocer las flores incluso en poco rotadas
*   Desplazamiento (vertical y horizontal): ambas generar imágenes con sentido y en casos hasta ayuda a centrar la flor. También ayuda a que el modelo identifique la flor sin importar su posición
*   Inclinación (shear): cambiar un poco la inclinación ayuda a reconocer desde distintos ángulos y sigue generando imágenes realistas (si no es en exceso)
*   Zoom: Ayuda a que el modelo no dependa del tamaño de la flor y cómo las anteriores genera imágenes realistas
*   Flip Horizontal: la flor volteada horizontalmente sigue siendo la misma flor y es realista. El vertical no tendría sentido

**Tarea 15: Personalizar el bloque de decisión**  
Añadimos una capa de flatten y tres nuevas capas densas especificas para nuestro problema con dimensiones 4096, 1072 y la que necesite la capa de salida (con sus correspondientes drop-outs).

```python
#Nuevo top (usando API funcional)
top_model = conv_base.output
top_model = keras.layers.Flatten()(top_model) #Añadimos la capa de aplanamiento
top_model = keras.layers.Dense(4096, activation='relu')(top_model)#Añadimos capa densa
top_model = keras.layers.Dense(1072, activation='relu')(top_model)#Añadimos capa densa
top_model = keras.layers.Dropout(0.5)(top_model)#Añadimos capa de drop out
output_layer = keras.layers.Dense(5, activation='softmax')(top_model)#Añadimos la capa de salida
```

**Tarea 17: Crear los datagenerators oportunos**

Para ello:
> * Utiliza Data augmentation.  
> * Las imágenes tienen que ser preprocesadas igual que cuando se entrenó el modelo pre.entrenado original. Para ello se utiliza el parámetro `preprocessing_function=preprocess_input` (preprocess_input importado desde `keras.applications.xxxxxx` en ambos generators (train y valid). Por tanto, no hay que indicarle `rescale`. En caso de ser necesario, se encargará `preprocess_input`.  
> * Dado que estamos reutilizando un modelo que no "es nuestro", deberemos ceñirnos al tamaño de imagen que permite la red a la entrada. Recuerda que debe ser: 224x224.

```python
from keras.applications.resnet50 import preprocess_input
#No hacemos el rescale y hacemos la función de preprocesado que se nos indica en el enunciado, previamente importada
#Aplicamos las técnincas de data augmentation como anteriormente.
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=10,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
   )

val_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)
```

```python
train_generator = train_datagen.flow_from_directory(
        train_dir,
        target_size=(224, 224), #El tamaño cambia respecto a los ejercicios anteriores, ya que lo tuvimos que cambiar en esta sección por requerimiento de nuestra red preentrenada
        batch_size=100,
        class_mode='sparse')

validation_generator = val_datagen.flow_from_directory(
        val_dir,
        target_size=(224, 224),
        batch_size=100,
        class_mode='sparse')
```

A la hora de entrenar un modelo pretrained es típico bajar el learning rate respecto al que utilizaríamos para un modelo nuestro desde cero.

```python
#Entrenamos el modelo igual que en los modelos anteriores
model_3 = keras.Model(inputs=conv_base.input, outputs=output_layer)

optim = keras.optimizers.Adam(learning_rate=0.0001) #Disminuimos el learning rate como se nos indica
model_3.compile(loss='SparseCategoricalCrossentropy',
              optimizer=optim,
              metrics=['acc'])

es_callback = keras.callbacks.EarlyStopping(patience=4, restore_best_weights=True)
history3 = model_3.fit(
      train_generator,
      steps_per_epoch=100,
      epochs=15,
      validation_data=validation_generator,
      validation_steps=50,
      callbacks=[es_callback],
      verbose=2)
```
